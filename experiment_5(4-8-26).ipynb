# Discretization and binning:

```python
data_for_cut = fraud_test_df['amt']
import pandas as pd
bins = [0, 50, 200, 1000, data_for_cut.max() + 1]
category = pd.cut(data_for_cut, bins)
print(category)
```

```python
print(pd.Series(category).value_counts())
```

```python
bin_names = ['Small Amount', 'Medium Amount', 'Large Amount', 'Very Large Amount']
print(pd.cut(data_for_cut, bins, labels=bin_names))
```

## Feature scaling and normalization:

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

data = {
 'Age': [25, 30, 45, 35, 22],
 'Salary': [50000, 64000, 120000, 85000, 1200000]
}
df = pd.DataFrame(data)
print("Original Data:\n", df, "\n")

min_max_scaler = MinMaxScaler()
df_min_max = pd.DataFrame(
 min_max_scaler.fit_transform(df),
 columns=df.columns
)
print("Min-Max Scaled Data:\n", df_min_max, "\n")

standard_scaler = StandardScaler()
df_standard = pd.DataFrame(
 standard_scaler.fit_transform(df),
 columns=df.columns
)
print("Standardized Data:\n", df_standard, "\n")
```
